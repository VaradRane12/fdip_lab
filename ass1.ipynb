```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img = cv2.imread('img.jpg', cv2.IMREAD_GRAYSCALE)
plt.imshow(img, cmap='gray')
plt.title('Original')
plt.axis('off')
plt.show()
```

```python
brightness_increase = cv2.convertScaleAbs(img, alpha=1, beta=50)
plt.imshow(brightness_increase, cmap='gray')
plt.title('Brightness Increased')
plt.axis('off')
plt.show()
```

```python
brightness_decrease = cv2.convertScaleAbs(img, alpha=1, beta=-50)
plt.imshow(brightness_decrease, cmap='gray')
plt.title('Brightness Decreased')
plt.axis('off')
plt.show()
```

```python
_, threshold_img = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)
plt.imshow(threshold_img, cmap='gray')
plt.title('Thresholding')
plt.axis('off')
plt.show()
```

```python
negative_img = 255 - img
plt.imshow(negative_img, cmap='gray')
plt.title('Negative')
plt.axis('off')
plt.show()
```

```python
img_float = img.astype(np.float32)
log_transformation = cv2.normalize(np.log1p(img_float), None, 0, 14, cv2.NORM_MINMAX)
log_transformation = np.uint8(log_transformation)

plt.imshow(log_transformation, cmap='gray')
plt.title('Log Transform')
plt.axis('off')
plt.show()
```

```python
gamma = 2.0
gamma_corrected = np.array(255 * (img / 255.0) ** gamma, dtype='uint8')
plt.imshow(gamma_corrected, cmap='gray')
plt.title(f'Power Law, Gamma: {gamma}')
plt.axis('off')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img = cv2.imread('img.jpg', cv2.IMREAD_GRAYSCALE)

brightness_increase = cv2.convertScaleAbs(img, alpha=1, beta=50)

brightness_decrease = cv2.convertScaleAbs(img, alpha=1, beta=-50)

_, threshold_img = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)

negative_img = 255 - img

img_float = img.astype(np.float32)
log_transformation = cv2.normalize(np.log1p(img_float), None, 0, 14, cv2.NORM_MINMAX)
log_transformation = np.uint8(log_transformation)

gamma = 2.0
gamma_corrected = np.array(255 * (img / 255.0) ** gamma, dtype='uint8')

titles = ['Original', 'Brightness Increased', 'Brightness Decreased',
          'Thresholding', 'Negative', 'Log Transform', f'Power Law, Gamma: {gamma}']
images = [img, brightness_increase, brightness_decrease,
          threshold_img, negative_img, log_transformation, gamma_corrected]

plt.figure(figsize=(14, 8))
for i in range(len(images)):
    plt.subplot(2, 4, i + 1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')
plt.tight_layout()
plt.savefig("output.png")
plt.show()
```
