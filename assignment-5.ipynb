```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img = cv2.imread("img.jpg", cv2.IMREAD_GRAYSCALE)
```

```python
def contrast_stretch(image):
    min_val = np.min(image)
    max_val = np.max(image)
    stretched = ((image - min_val) / (max_val - min_val)) * 255
    return stretched.astype(np.uint8)
```

```python
contrast_img = contrast_stretch(img)
equalized_img = cv2.equalizeHist(img)

fig, axs = plt.subplots(2, 3, figsize=(15, 8))

axs[0,0].imshow(img, cmap='gray')
axs[0,0].set_title("Original Image")
axs[0,0].axis("off")
axs[1,0].hist(img.ravel(), bins=256, range=[0,256], color='black')
axs[1,0].set_title("Histogram - Original")

axs[0,1].imshow(contrast_img, cmap='gray')
axs[0,1].set_title("Contrast Stretched")
axs[0,1].axis("off")
axs[1,1].hist(contrast_img.ravel(), bins=256, range=[0,256], color='black')
axs[1,1].set_title("Histogram - Contrast Stretch")

axs[0,2].imshow(equalized_img, cmap='gray')
axs[0,2].set_title("Histogram Equalized")
axs[0,2].axis("off")
axs[1,2].hist(equalized_img.ravel(), bins=256, range=[0,256], color='black')
axs[1,2].set_title("Histogram - Equalized")

plt.tight_layout()
plt.savefig("output.png")
plt.show()
```
